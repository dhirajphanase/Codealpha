```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("unemployment_data.csv")
```

```python
print(df.head())

df.columns = df.columns.str.strip().str.lower()
print("Cleaned Columns:", df.columns.tolist())

df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date')
```

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='date', y='estimated unemployment rate', marker='o')
plt.title('Unemployment Rate Over Time (Focus on COVID-19 Period)', fontsize=16)
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.axvspan(pd.to_datetime('2020-03-01'), pd.to_datetime('2021-03-01'), color='red', alpha=0.2, label='COVID-19 Peak')
plt.legend()
plt.tight_layout()
plt.show()
```
